```python
import pandas as pd
import numpy as np

df = pd.read_excel("BankChurners.xlsx")
df.shape
```

```python
# Step 1: Drop Naive Bayes leakage columns
nb_cols = [c for c in df.columns if c.startswith('Naive_Bayes')]
df.drop(columns=nb_cols, inplace=True)

print(f"Columns remaining: {df.shape[1]}")
print(df.columns.tolist())
```

```python
# Step 2: Replace "Unknown" with NaN (encoded missing values)
df = df.replace("Unknown", np.nan)

# Verify
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])
```

```python
# Step 3: Set CLIENTNUM as index
df.set_index('CLIENTNUM', inplace=True)
print(df.index.name)
print(df.shape)
```

```python
income_order = ['Less than $40K', '$40K - $60K', '$60K - $80K',
                '$80K - $120K', '$120K +']
df['Income_Category'] = pd.Categorical(
    df['Income_Category'], categories=income_order, ordered=True)
print(df['Income_Category'].dtype)
```

```python
card_order = ['Blue', 'Silver', 'Gold', 'Platinum']
df['Card_Category'] = pd.Categorical(
    df['Card_Category'], categories=card_order, ordered=True)
print(df['Card_Category'].dtype)
print(df['Card_Category'].cat.categories.tolist())
print(df['Card_Category'].value_counts())
```

```python
df['Churned'] = df['Attrition_Flag'].map(
    {'Existing Customer': 0, 'Attrited Customer': 1})
df.drop(columns=['Attrition_Flag'], inplace=True)
print(df['Churned'].value_counts())
```

```python
print(df.shape)
print(df['Income_Category'].dtype)
print(df['Card_Category'].dtype)
print(df['Churned'].dtype)
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])
df.head(3)
```

```python
df.to_csv("bankchurners_clean.csv", index=True)
print("Saved.")
```
